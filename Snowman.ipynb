```python
import random

def choose_word(language, difficulty):
  if language == 1:
    if difficulty == 1:
      wordbankeasy = ['apple', 'ships', 'agree', 'enjoy', 'model', 'music', 'loyal', 'hello']
      choice_easy =  random.choice(wordbankeasy)
      return choice_easy
    else:
      wordbankhard = ['appreciate', 'friendship', 'motivation', 'restaurant', 'earthquake', 'government']
      choice_hard = random.choice(wordbankhard)
      return choice_hard
  elif language == 2:
    if difficulty == 1:
      wordbankeasy = ['perro', 'libro', 'clase', 'tener', 'vivir', 'comer', 'reloj']
      choice_easy = random.choice(wordbankeasy)
      return choice_easy
    else:
      wordbankhard = ['cumpleanos', 'estudiante', 'importante', 'dedicacion', 'transporte', 'cientifico']
      choice_hard = random.choice(wordbankhard)
      return choice_hard
#I used google to tell me how to get random decisions

def snowman(word, guesses):
  storedword =  ""
  wordlist = list(word)
  for i in word:
    storedword += "-"
  for i in range(guesses + 1):
    if word != storedword and guesses > 0:
      print(f'You have {guesses} guesses left ')
      guess = input('Give me your guess, if you would like a hint enter hint ')
      if guess == 'hint':
            storedword = "".join(wordlist[:2:]) + storedword[2::]
      else:
        for x in range(len(wordlist)):
            if wordlist[x] == guess:
              storedword = storedword[:x] + guess + storedword[x + 1::]
            elif word == guess:
              storedword = word
            elif len(guess) > 1 and guess != word:
              guesses = 0
              break
            else:
              pass
    elif word == storedword:
       print(f'You got the hidden word! It was {word}!')
       break
    else:
      print(f'You ran out of guesses, the hidden word was {word} ')
      break
    guesses -= 1
    print(storedword)

#storedword = "".join(wordlist[:1:])

print('Welcome to Snowman! You will have 10 guesses to find the word.')
lan = int(input('Would you like the word to be in english or spanish. For english press 1, for spanish press 2. '))
while True:
  try:
    dif = int(input('If you want the difficulty to be hard press 2. For Easy press 1. '))
    break
  except ValueError:
    print('That isnt one of the options')
snowman(choose_word(lan, dif), 10)
#Third special feature: bonus round
bonus = int(input('Would you like to play a bonus round with only half the guesses? For yes press (1). For no press (2). '))
if bonus == 1:
  snowman(choose_word(lan, dif), 5)
else:
  print('Have a good day!')
```
